```python
import pickle
import pandas as pd

# Load model
with open("ml3_output/ml3_best_model.pkl", "rb") as f:
    model_package = pickle.load(f)

model = model_package["model"]
features = model_package["features"]

# New observation
new_data = pd.DataFrame([{
    "avg_activity": 690.5,
    "activity_growth": 0.12,
    "active_hours": 18,
    "peak_ratio": 1.45,
    "variability": 1.0,
    "internet_share": 0.72
}])

# Ensure correct feature order
new_data = new_data[features]

# Prediction
prediction = model.predict(new_data)[0]

print("HIGH_ACTIVITY =", prediction)

if prediction == 1:
    print("Predicted: High Activity Period")
else:
    print("Predicted: Normal Activity Period")
```

```python
probability = model.predict_proba(new_data)[0][1]

print(f"Probability of HIGH_ACTIVITY: {probability:.4f}")
```
